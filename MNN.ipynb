```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Размеры входной и выходной матриц
input_shape = (10, 20)  # Входная матрица 10x20
output_shape = (5, 15)  # Выходная матрица 5x15

# Модель
model = models.Sequential([
    layers.Flatten(input_shape=input_shape),  # Разворачивание матрицы
    layers.Dense(output_shape[0] * output_shape[1]),  # Полносвязный слой
    layers.Reshape(output_shape)  # Преобразование в матрицу
])

# Проверка модели
input_data = tf.random.normal((32, 10, 20))  # Пакет из 32 матриц
output_data = model(input_data)
print(output_data.shape)  # Ожидается (32, 5, 15)
```

создаёте тензор формы ( 𝑏 𝑎 𝑡 𝑐 ℎ _ 𝑠 𝑖 𝑧 𝑒 , 𝑚 , 𝑛 , 2 ) ГДЕ batch_size — размер пакета данных,m и 
𝑛 n — размеры матрицы,2 — количество чисел в одной ячейке.


```python
import tensorflow as tf

# Размер матрицы и пакета
batch_size = 32
matrix_height = 10
matrix_width = 20

# Создание случайного тензора
input_tensor = tf.random.normal((batch_size, matrix_height, matrix_width, 2))
print(input_tensor.shape)  # Ожидается: (32, 10, 20, 2)
```

2.1 Свёрточные слои
Свёрточные слои могут обрабатывать тензоры с дополнительным измерением, рассматривая его как количество каналов. Например, если в каждой ячейке 2 числа, они будут обрабатываться как двухканальное изображение.

```python
from tensorflow.keras import layers, models
```

```python

# Создание модели
model = models.Sequential([
    layers.Conv2D(16, (3, 3), activation='relu', input_shape=(10, 20, 2)),  # Свёрточный слой
    layers.Conv2D(8, (3, 3), activation='relu'),  # Ещё один свёрточный слой
    layers.Flatten(),  # Разворачивание вектора
    layers.Dense(10)  # Полносвязный слой
])

# Проверка модели
output_tensor = model(input_tensor)
print(output_tensor.shape)  # Ожидается: (32, 10)
```

2.2 Полносвязные слои
Если вы хотите преобразовать тензор в другой тензор той же формы, можно использовать полносвязные слои с функцией активации.

```python
model = models.Sequential([
    layers.Flatten(input_shape=(10, 20, 2)),  # Разворачивание тензора
    layers.Dense(10 * 20 * 2, activation='relu'),  # Полносвязный слой
    layers.Reshape((10, 20, 2))  # Возврат к исходной форме
])
```

3. Обучение с такими данными

```python
# Генерация данных
X_train = tf.random.normal((100, 10, 20, 2))  # 100 входных тензоров
Y_train = tf.random.normal((100, 10, 20, 2))  # 100 целевых тензоров

# Компиляция модели
model.compile(optimizer='adam', loss='mse')

# Обучение
model.fit(X_train, Y_train, epochs=10, batch_size=16)
```
